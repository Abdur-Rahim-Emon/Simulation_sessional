```python
import random
import math
import matplotlib.pyplot as plt
#Show graphs directly inside Jupyter Notebook.
%matplotlib inline 
```

```python
point_counts = [100, 1000, 10000, 1000000]
pi_estimates = []
errors = []

random.seed(42)

for n in point_counts:
    h = 0  # Points inside the quarter-circle
    
    for i in range(n):
        x = random.uniform(0, 1)
        y = random.uniform(0, 1)
        
        if (x * x) + (y * y) <= 1:
            h = h + 1
            
    estimated_pi = (4 * h) / n
    error = abs(math.pi - estimated_pi)
    
    pi_estimates.append(estimated_pi)
    errors.append(error)

print("Simulation completed!")
```

```python
print("-" * 55)
print(f"{'Points (n)':<15} | {'Estimated Pi':<15} | {'Absolute Error':<15}")
print("-" * 55)

for i in range(len(point_counts)):
    print(f"{point_counts[i]:<15} | {pi_estimates[i]:<15.6f} | {errors[i]:<15.6f}")
print("-" * 55)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(point_counts, pi_estimates, marker='o', linestyle='-', color='blue', label='Estimated Pi')
plt.axhline(y=math.pi, color='red', linestyle='--', label='Actual Pi (3.14159...)')

plt.xscale('log')
plt.title('Monte Carlo Estimation of Pi vs. Number of Points')
plt.xlabel('Number of Generated Points (Log Scale)')
plt.ylabel('Estimated Value of Pi')
plt.legend()
plt.grid(True)
plt.show()
```
