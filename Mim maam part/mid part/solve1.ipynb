```python
import numpy as np

# Generate 100 random numbers uniformly distributed between 0 and 1
random_numbers = np.random.rand(1000)

# Calculate mean, minimum, and standard deviation
mean_val = np.mean(random_numbers)
min_val = np.min(random_numbers)
std_val = np.std(random_numbers)

print(f"Mean: {mean_val:.4f}")
print(f"Minimum: {min_val:.4f}")
print(f"Standard Deviation: {std_val:.4f}")
```

```python
# Generate another set of 100 random numbers without setting a seed
random_numbers_2 = np.random.rand(100)
mean_val_2 = np.mean(random_numbers_2)

print(f"First Mean:  {mean_val:.4f}")
print(f"Second Mean: {mean_val_2:.4f}")
```

```python
# Set a fixed random seed (e.g., 42) for reproducibility
np.random.seed(42)
run_1 = np.random.rand(100)

# Reset the seed to the exact same value before generating again
np.random.seed(42)
run_2 = np.random.rand(100)

print(f"Run 1 Mean: {np.mean(run_1):.4f}")
print(f"Run 2 Mean: {np.mean(run_2):.4f}")
print(f"Are both arrays identical? {np.array_equal(run_1, run_2)}")
"""  standard mean of random numbers 0 to 1 is 0.5 
and  standard deviation is b-a/12 = 
"""
```
