```python
import numpy as np
import matplotlib.pyplot as plt


# for gaining same result seed is used 
np.random.seed(42)

n = 1000  # customer number

# ১. ১০০০ Inter-arrival time crating (Exponential distribution, mean = 5)
mean_inter_arrival = 5.0
inter_arrival_times = np.random.exponential(scale=mean_inter_arrival, size=n)
""" printing inter arrival time code 
print("inter-arrival time  \n")
print(inter_arrival_times)
"""
# ২. ১০০০  Service time create  (Uniform distribution, 2 to 8 )
low = 2.0
high = 8.0
service_times = np.random.uniform(low=low, high=high, size=n)

print("1000 Inter-arrival times and Service times generated successfully!")
```

```python
%matplotlib inline

plt.figure(figsize=(12, 5))

# 1st graph: Inter-arrival times
plt.subplot(1, 2, 1)
plt.hist(inter_arrival_times, bins=30, color='skyblue', edgecolor='black')
plt.title('Inter-Arrival Times (Exponential)')
plt.xlabel('Time (minutes)')
plt.ylabel('Number of Customers')

# 2nd graph: Service times
plt.subplot(1, 2, 2)
plt.hist(service_times, bins=30, color='lightgreen', edgecolor='black')
plt.title('Service Times (Uniform)')
plt.xlabel('Time (minutes)')
plt.ylabel('Number of Customers')

plt.tight_layout()
plt.show()
```

```python
# Inter-arrival time- calculation
sample_mean_ia = np.mean(inter_arrival_times)
sample_std_ia = np.std(inter_arrival_times, ddof=1) #s = √( Σ(xᵢ − x̄)² / (n − 1) )
#ddof = Delta Degrees of Freedom
# Service time calculation 
sample_mean_st = np.mean(service_times)
sample_std_st = np.std(service_times, ddof=1)

print(f"Inter-Arrival -> Sample Mean: {sample_mean_ia:.2f}, Sample Std Dev: {sample_std_ia:.2f}")
print(f"Service Time  -> Sample Mean: {sample_mean_st:.2f}, Sample Std Dev: {sample_std_st:.2f}")
```

```python
# Theoretical calculation part 
theo_mean_ia = mean_inter_arrival
theo_std_ia = mean_inter_arrival

theo_mean_st = (low + high) / 2
theo_std_st = (high - low) / np.sqrt(12)

# Analysis part 
print("=== INTER-ARRIVAL TIME COMPARISON ===")
print(f"Sample Mean:      {sample_mean_ia:.2f}  |  Theoretical Mean:    {theo_mean_ia:.2f}")
print(f"Sample Std Dev:   {sample_std_ia:.2f}  |  Theoretical Std Dev: {theo_std_ia:.2f}")

print("\n=== SERVICE TIME COMPARISON ===")
print(f"Sample Mean:      {sample_mean_st:.2f}  |  Theoretical Mean:    {theo_mean_st:.2f}")
print(f"Sample Std Dev:   {sample_std_st:.2f}  |  Theoretical Std Dev: {theo_std_st:.2f}")
```
